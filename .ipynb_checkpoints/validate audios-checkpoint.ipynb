```python

import pandas as pd
import numpy as np
import soundfile as sf
from pathlib import Path

# Project directories
PROJECT_DIR = Path(".").resolve()
INPUT_DIR = PROJECT_DIR / "inputs"
OUTPUT_DIR = PROJECT_DIR / "outputs"

OUTPUT_DIR.mkdir(exist_ok=True)

# Load manifest
manifest = pd.read_csv(INPUT_DIR / "manifest.csv")

results = []

for _, row in manifest.iterrows():

    audio_path = INPUT_DIR / row["audio_path"]

    reference_path = INPUT_DIR / row["reference_audio_path"]

    record = {
        "clip_id": row["clip_id"],
        "language": row["language"],
        "audio_exists": audio_path.exists(),
        "reference_exists": reference_path.exists()
    }

    # Validate main audio
    if audio_path.exists():
        try:
            info = sf.info(str(audio_path))
            audio, sr = sf.read(str(audio_path))

            record["sample_rate"] = sr
            record["channels"] = info.channels
            record["duration"] = info.duration
            record["silent"] = (
                np.max(np.abs(audio)) < 1e-5
            )

            record["audio_valid"] = (
                info.frames > 0
                and not record["silent"]
            )

        except Exception as e:
            record["audio_valid"] = False
            record["audio_error"] = str(e)

    else:
        record["audio_valid"] = False

    # Validate reference audio
    if reference_path.exists():
        try:
            ref_info = sf.info(str(reference_path))

            record["reference_duration"] = ref_info.duration
            record["reference_sample_rate"] = (
                ref_info.samplerate
            )
            record["reference_valid"] = (
                ref_info.frames > 0
            )

        except Exception as e:
            record["reference_valid"] = False
            record["reference_error"] = str(e)

    else:
        record["reference_valid"] = False

    results.append(record)


# Save validation report
report = pd.DataFrame(results)

report.to_csv(
    OUTPUT_DIR / "audio_validation.csv",
    index=False
)

# Display results
print("\n--- AUDIO VALIDATION ---")
print(report["audio_valid"].value_counts())

print("\n--- REFERENCE VALIDATION ---")
print(report["reference_valid"].value_counts())

print("\n--- MISSING AUDIO ---")
print(
    report.loc[
        ~report["audio_exists"], "clip_id"
    ].tolist()
)

print("\n--- MISSING REFERENCES ---")
print(
    report.loc[
        ~report["reference_exists"], "clip_id"
    ].tolist()
)

print("\n--- SAMPLING RATES ---")
print(report["sample_rate"].value_counts())

print("\n--- LANGUAGE DISTRIBUTION ---")
print(report.groupby("language")["clip_id"].count())

print("\nReport saved successfully!")
```

```python
import torch

results = []

for _, row in pilot_df.iterrows():
    audio_path = PROJECT_DIR / "inputs" / row["audio_path"]

    try:
        # Load waveform
        waveform, sr = torchaudio.load(str(audio_path))

        # Convert to mono if necessary
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0, keepdim=True)

        # Resample if needed
        if sr != 16000:
            waveform = torchaudio.functional.resample(
                waveform,
                orig_freq=sr,
                new_freq=16000
            )

        waveform = waveform.float().to(device)

        # Score one clip at a time
        with torch.no_grad():
            score = utmos.score(waveform)

        results.append({
            "clip_id": row["clip_id"],
            "language": row["language"],
            "audio_path": row["audio_path"],
            "utmos_score": float(score.item()),
            "status": "success"
        })

    except Exception as e:
        results.append({
            "clip_id": row["clip_id"],
            "language": row["language"],
            "audio_path": row["audio_path"],
            "utmos_score": None,
            "status": str(e)
        })

    print(f"Processed: {row['clip_id']}")

utmos_pilot = pd.DataFrame(results)

display(utmos_pilot)
```

```python
import torch
import torchaudio
import numpy as np
import pandas as pd

from transformers import AutoFeatureExtractor, AutoModel
from pathlib import Path

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
```

```python
MODEL_NAME = "facebook/wav2vec2-xls-r-300m"

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

feature_extractor = AutoFeatureExtractor.from_pretrained(
    MODEL_NAME
)

xlsr_model = AutoModel.from_pretrained(
    MODEL_NAME
)

xlsr_model = xlsr_model.to(device)
xlsr_model.eval()

print("XLS-R loaded successfully!")
print("Device:", device)
print("Hidden size:", xlsr_model.config.hidden_size)
```

```python
from pathlib import Path
import pandas as pd
import numpy as np
import torch
import torchaudio

PROJECT_DIR = Path.cwd()

manifest_path = PROJECT_DIR / "inputs" / "manifest.csv"
manifest = pd.read_csv(manifest_path)

features_dir = PROJECT_DIR / "features"
features_dir.mkdir(exist_ok=True)

print("Project directory:", PROJECT_DIR)
print("Manifest shape:", manifest.shape)
print("Features directory:", features_dir)
```

```python
pilot_df = pd.concat([
    manifest[manifest["language"] == "hi"].head(4),
    manifest[manifest["language"] == "ta"].head(4)
]).reset_index(drop=True)

display(pilot_df[["clip_id", "language", "audio_path"]])
```

```python
# Select the first Hindi clip
sample_row = manifest.iloc[0]

audio_path = PROJECT_DIR / "inputs" / sample_row["audio_path"]

waveform, sr = torchaudio.load(str(audio_path))

# Convert to mono
if waveform.shape[0] > 1:
    waveform = waveform.mean(dim=0)

# Resample if required
if sr != 16000:
    waveform = torchaudio.functional.resample(
        waveform,
        orig_freq=sr,
        new_freq=16000
    )

audio_array = waveform.numpy()

inputs = feature_extractor(
    audio_array,
    sampling_rate=16000,
    return_tensors="pt"
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

with torch.no_grad():
    outputs = xlsr_model(**inputs)

print("Last hidden state shape:", outputs.last_hidden_state.shape)
```

```python
sample_embedding = outputs.last_hidden_state.mean(dim=1)

sample_embedding = sample_embedding.squeeze(0).cpu().numpy()

print("Embedding shape:", sample_embedding.shape)
```

```python
pilot_embeddings = []

for _, row in pilot_df.iterrows():

    audio_path = PROJECT_DIR / "inputs" / row["audio_path"]

    try:
        waveform, sr = torchaudio.load(str(audio_path))

        # Convert to mono
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0)

        # Ensure 16 kHz
        if sr != 16000:
            waveform = torchaudio.functional.resample(
                waveform,
                orig_freq=sr,
                new_freq=16000
            )

        audio_array = waveform.numpy()

        inputs = feature_extractor(
            audio_array,
            sampling_rate=16000,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        with torch.no_grad():
            outputs = xlsr_model(**inputs)

        embedding = (
            outputs.last_hidden_state
            .mean(dim=1)
            .squeeze(0)
            .cpu()
            .numpy()
        )

        result = {
            "clip_id": row["clip_id"],
            "language": row["language"],
            "audio_path": row["audio_path"]
        }

        for i, value in enumerate(embedding):
            result[f"xlsr_{i:04d}"] = float(value)

        pilot_embeddings.append(result)

        print("Processed:", row["clip_id"])

    except Exception as e:
        print("Failed:", row["clip_id"], "|", str(e))
```

```python
xlsr_pilot_df = pd.DataFrame(pilot_embeddings)

xlsr_pilot_path = "feature mapping/xlsr_features_pilot.csv"

xlsr_pilot_df.to_csv(xlsr_pilot_path, index=False)

print("Successful clips:", len(xlsr_pilot_df))
print("Total columns:", len(xlsr_pilot_df.columns))
print(
    "Embedding dimensions:",
    len([c for c in xlsr_pilot_df.columns if c.startswith("xlsr_")])
)

print("Saved:", xlsr_pilot_path)
```

```python
embedding_cols = [
    c for c in xlsr_pilot_df.columns
    if c.startswith("xlsr_")
]

print("Missing values:", xlsr_pilot_df[embedding_cols].isna().sum().sum())
print("Infinite values:", np.isinf(xlsr_pilot_df[embedding_cols].values).sum())

display(xlsr_pilot_df.iloc[:, :10])
```

```python
xlsr_features_all = []

for idx, row in manifest.iterrows():

    audio_path = PROJECT_DIR / "inputs" / row["audio_path"]

    try:
        waveform, sr = torchaudio.load(str(audio_path))

        # Convert to mono
        if waveform.shape[0] > 1:
            waveform = waveform.mean(dim=0)

        # Resample to 16 kHz
        if sr != 16000:
            waveform = torchaudio.functional.resample(
                waveform,
                orig_freq=sr,
                new_freq=16000
            )

        audio_array = waveform.numpy()

        inputs = feature_extractor(
            audio_array,
            sampling_rate=16000,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        with torch.no_grad():
            outputs = xlsr_model(**inputs)

        embedding = (
            outputs.last_hidden_state
            .mean(dim=1)
            .squeeze(0)
            .cpu()
            .numpy()
        )

        result = {
            "clip_id": row["clip_id"],
            "language": row["language"],
            "audio_path": row["audio_path"]
        }

        for i, value in enumerate(embedding):
            result[f"xlsr_{i:04d}"] = float(value)

        xlsr_features_all.append(result)

    except Exception as e:
        print(f"Failed: {row['clip_id']} | {e}")

    if (idx + 1) % 20 == 0:
        print(f"Processed {idx + 1}/{len(manifest)} clips")
```

```python
xlsr_features_df = pd.DataFrame(xlsr_features_all)

xlsr_output = "feature mapping/xlsr_features.csv"

xlsr_features_df.to_csv(xlsr_output, index=False)

print("Successful clips:", len(xlsr_features_df))
print("Feature dimensions:", len([
    c for c in xlsr_features_df.columns
    if c.startswith("xlsr_")
]))

print("Saved:", xlsr_output)
```

```python
embedding_cols = [
    c for c in xlsr_features_df.columns
    if c.startswith("xlsr_")
]

print("Missing values:",
      xlsr_features_df[embedding_cols].isna().sum().sum())

print("Infinite values:",
      np.isinf(xlsr_features_df[embedding_cols].values).sum())

print("Duplicate clip IDs:",
      xlsr_features_df["clip_id"].duplicated().sum())

print("Language distribution:")
display(xlsr_features_df["language"].value_counts())
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.cwd()
features_dir = PROJECT_DIR / "features"

utmos_df = pd.read_csv(
    "feature mapping/utmos_features.csv"
)

xlsr_df = pd.read_csv(
    "feature mapping/xlsr_features.csv"
)

print("UTMOS shape:", utmos_df.shape)
print("XLS-R shape:", xlsr_df.shape)
```

```python
# Keep UTMOS metadata and score
utmos_selected = utmos_df[
    ["clip_id", "language", "audio_path", "utmos_score"]
].copy()

# Select all 1,024 XLS-R dimensions
xlsr_columns = [
    col for col in xlsr_df.columns
    if col.startswith("xlsr_")
]

xlsr_selected = xlsr_df[
    ["clip_id"] + xlsr_columns
].copy()

print("UTMOS columns:", len(utmos_selected.columns))
print("XLS-R dimensions:", len(xlsr_columns))
```

```python
naturalness_features = pd.merge(
    utmos_selected,
    xlsr_selected,
    on="clip_id",
    how="inner",
    validate="one_to_one"
)

print("Combined dataset shape:", naturalness_features.shape)

display(naturalness_features.head())
```

```python
print("Total clips:", len(naturalness_features))

print(
    "Unique clip IDs:",
    naturalness_features["clip_id"].nunique()
)

print(
    "Missing values:",
    naturalness_features.isna().sum().sum()
)

feature_cols = ["utmos_score"] + xlsr_columns

print(
    "Infinite feature values:",
    np.isinf(
        naturalness_features[feature_cols].to_numpy()
    ).sum()
)

print("\nLanguage distribution:")
display(naturalness_features["language"].value_counts())
```

```python
output_path = (
    features_dir / "naturalness_combined_features.csv"
)

naturalness_features.to_csv(
    output_path,
    index=False
)

print("Saved successfully:", output_path)
```
